## Validation Strategy

A row-level random split would be misleading because repeated observations of the same `legId` could appear in train and validation/test. The project uses grouped splitting by `legId` so every itinerary belongs to exactly one split.

## BOOK/WAIT Decision Support

BOOK/WAIT is framed as a historical decision-support experiment:

`WAIT = 1` if the same `legId` had a fare at least $10 cheaper within the next 7 search days.

This is more valid than using only `days_left`, but it still does not guarantee future live fare movement.
